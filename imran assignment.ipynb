## ARITHMETIC OPERATORS

```python
a=10
b=15
print(f"Addition of a and b is: {a+b}")
print(f"Subtraction of a and b is: {a-b}")
print(f"Multiplication of a and b is: {a*b}")
print(f"Division of a and b is: {a/b}") 
print(f"Modulus of a and b is: {a%b}")
print(f"Floor division of a and b is: {a//b}")
print(f"Exponentiation of a and b is: {a**b}")
```

## COMPARISION OR RELATIONAL OPERATORS

```python
score = 85
print(score >= 100)
print(score == 100)
print(score <= 100)
print(score != 100)
print(score > 100)
print(score < 100)
```

## LOGICAL OPERATORS

```python
a=75
b=87
print(a<b and a>b)
print(a<b or a>b)
print(not(a<b))
```

## BITWISE OPERATORS

```python
a=6
b=9
print(a&b)
print(a|b)
print(a^b)
print(~a)
print(a<<1)
print(a>>1)
print(b<<1)
print(b>>1)
```

## ASSIGNMENT OPERATORS

```python
x=10
print(x)

x -= 5
print(x)

x += 5
print(x)

x *= 5
print(x)

x /= 5 
print(x)

x //= 5
print(x)

x = int(x)
print(x)

x |= 5
print(x)

x ^= 5
print(x)

x <<= 5
print(x)

x >>= 5
print(x)

```
